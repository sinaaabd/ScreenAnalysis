```python
import pandas as pd
import matplotlib.pyplot as plt

# Cluster_number -> cluster_id
```

```python
df = pd.read_csv("data/Hit_grouped_final.csv")
```

```python
plt.hist(df['cluster_id'], bins=len(df['cluster_id'].unique()))
plt.xlabel("Cluster ID")
plt.ylabel("Frequency")
plt.show()
```

```python
top_clusters = [3068, 3837, 3534]
df['cluster_id'].value_counts().sort_values().tail(5)
```

```python
# Get ligands in top clusters
df_wt = df.loc[df['Protein'] == 'PAH-WT']
df_mut = df.loc[df['Protein'] == 'PAH-I65T']

for c in top_clusters:
    print(f"cluster number: {c}")
    print(f"total number : {(df['cluster_id'] == c).sum()}")
    print(f"total number : {(df['cluster_id'] == c).sum()}")
    print('')
    print('WT')
    print(f"total number WT : {(df_wt['cluster_id'] == c).sum()}")
    count_table = df_wt.loc[df_wt['cluster_id'] == c, 'hit_group'].value_counts()
    print(count_table)
    df_to_save = pd.DataFrame({
        'cluster_id' : df_wt['cluster_id'] == c,
        'compound_id' : df_wt['ID']
    })
    df_to_save.to_excel(f"Hit_grouped_WT_cluster_{c}.xlsx")
    
    print('')
    print('I65T')
    print(f"total number I65T : {(df_mut['cluster_id'] == c).sum()}")
    count_table = df_mut.loc[df_mut['cluster_id'] == c, 'hit_group'].value_counts()
    print(count_table)
    df_to_save = pd.DataFrame({
        'cluster_id' : df_mut['cluster_id'] == c,
        'compound_id' : df_mut['ID']
    })
    df_to_save.to_excel(f"Hit_grouped_I65T_cluster_{c}.xlsx")
    print("----------")
```
