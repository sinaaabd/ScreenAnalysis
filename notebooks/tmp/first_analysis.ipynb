```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_main = pd.read_excel("../data/analysis_data/PO-ini2810.xlsx")
```

```python
df_main['Protein'].unique()
```

```python
protein = 'PAH-WT''

df = df_main.copy()
df = df.loc[df['Protein'] == protein].copy()
df = df.dropna(subset = ['Thermal_Shift'])
df = df.dropna(subset = ['Thermal_Shift'])
```

```python
df.groupby('ID').count()
```

```python
x = pd.to_numeric(df['delta_Tm_median'], errors='coerce')
y = pd.to_numeric(df['delta_SpSh_median'], errors='coerce')
y = np.abs(y)
np.corrcoef(x, y)[0,1]
```

```python
plt.hist(y, 500);
plt.xlim([0, .5]);
plt.title(protein);
```

```python
plt.hist(y, 500);
plt.xlim([0, .5]);
plt.title(protein);
```

```python
plt.scatter(x, y, s=.1, color='gray')
good = (x>3) & (y>.15)
ctrl = df['ID'].isin(['BH4', 'UCR102', 'UCR243'])
plt.scatter(x[good], y[good], s=1, color='red')
plt.scatter(x[ctrl], y[ctrl], s=1, color='blue')

# for i in range(len(good)):
#     if good[i]:
#         name = df.iloc[i]['ID']
#         plt.text(x[i], y[i], name)

plt.ylabel('delta_SpSh')
plt.xlabel('delta_Tm')
plt.title(protein)
plt.ylim([-.01, .6])
```
