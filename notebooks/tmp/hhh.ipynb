```python
import pandas as pd
```

```python
df = pd.read_csv("data/PO_norm_analysed_16may.csv", index_col = 0)
df = df.loc[~(df['ID'].isin(['BH4', 'UCR243', 'UCR102']))]
df = df.loc[df['Binding_validated_by_TSA'] == 'strong_evidence']
df.head()
```

```python
df['Protein'].unique()
```

```python
bind_id_wt = df.loc[df['Protein'] == 'PAH-WT', 'ID']
bind_id_mut = df.loc[df['Protein'] == 'PAH-I65T', 'ID']
# bind_id_wt, bind_id_mut
```

```python
intersection = list(set(bind_id_wt).intersection(set(bind_id_mut)))
```

```python
len(intersection)
```

```python
df_intersection = pd.DataFrame({'intersection'  : intersection})
df_intersection
```
